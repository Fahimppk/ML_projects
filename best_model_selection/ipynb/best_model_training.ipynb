```python
x=df.drop(columns=["Personal.Loan"])
y=df["Personal.Loan"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_LogisticRegression=LogisticRegression()
model_LogisticRegression.fit(x_train,y_train)
predict_LogisticRegression=model_LogisticRegression.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
accuracy_score_of_LogisticRegression=accuracy_score(y_test,predict_LogisticRegression)
print(accuracy_score_of_LogisticRegression)
```

```python
confusion_matrix_LogisticRegression=confusion_matrix(y_test,predict_LogisticRegression)
print(confusion_matrix_LogisticRegression)
```

```python
classification_report_LogisticRegression=classification_report(y_test,predict_LogisticRegression)
print(classification_report_LogisticRegression)
```

```python
sns.heatmap(confusion_matrix_LogisticRegression,annot=True,cmap="coolwarm",fmt="d")
plt.title("confusion_matrix - logical regression")
plt.xlabel("Logistic_model_predicted")
plt.ylabel("actual_value")
plt.show()
```

```python
model_names = [
    "Logistic Regression",
    "KNN",
    "SVC",
    "Naive Bayes",
    "Decision Tree",
    "AdaBoostClassifier",
    "RandomForestClassifier",
    "XGBClassifier",
    "GradientBoostingClassifier"
]

accuracies = [
    accuracy_score_of_LogisticRegression,
    accuracy_score_of_k_neighbors,
    accuracy_score_of_svm,
    accuracy_score_of_naive_bayes,
    accuracy_score_of_DecisionTree,
    accuracy_score_of_AdaBoost,
    accuracy_score_of_RandomForest,
    accuracy_score_of_XGBoost,
    accuracy_score_of_GradientBoost
]

accuracy_df = pd.DataFrame({
    "Model": model_names,
    "Accuracy": accuracies
})

accuracy_df = accuracy_df.sort_values(
    by="Accuracy",
    ascending=False
)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=accuracy_df,
    x="Model",
    y="Accuracy"
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.3f",
        padding=5
    )

plt.title("Accuracy Comparison of ML Models")
plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy Score")

plt.ylim(0, 1.1)
plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```
