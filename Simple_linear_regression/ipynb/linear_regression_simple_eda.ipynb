## Simple Exploratory Data Analysis (EDA)

These steps use basic Pandas, Matplotlib, and Seaborn commands. Each plot is written separately so it is easy to understand.

```python
# 1. Check the dataset
print(df.head())
print("Shape:", df.shape)
print(df.info())
```

```python
# 2. Statistical summary
df.describe()
```

```python
# 3. Check missing values
df.isnull().sum()
```

```python
# 4. Check duplicate rows
df.duplicated().sum()
```

```python
# 5. Check column data types
df.dtypes
```

```python
# 6. Histogram - YearsExperience
plt.figure(figsize=(6, 4))
sns.histplot(df['YearsExperience'], kde=True)
plt.title('Distribution of YearsExperience')
plt.show()
```

```python
# 7. Histogram - Salary
plt.figure(figsize=(6, 4))
sns.histplot(df['Salary'], kde=True)
plt.title('Distribution of Salary')
plt.show()
```

```python
# 8. Boxplot - YearsExperience
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['YearsExperience'])
plt.title('YearsExperience Boxplot')
plt.show()
```

```python
# 9. Boxplot - Salary
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Salary'])
plt.title('Salary Boxplot')
plt.show()
```

```python
# 10. Scatter plot - Experience vs Salary
plt.figure(figsize=(6, 4))
sns.scatterplot(x='YearsExperience', y='Salary', data=df)
plt.title('Years of Experience vs Salary')
plt.show()
```

```python
# 11. Regression plot - Experience vs Salary
plt.figure(figsize=(6, 4))
sns.regplot(x='YearsExperience', y='Salary', data=df)
plt.title('Regression Plot')
plt.show()
```

```python
# 12. Correlation
df.corr(numeric_only=True)
```

```python
# 13. Correlation heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
x=df.drop(columns='Salary')
```

```python
y=df['Salary']
```

```python
x.shape
```

```python
y.shape
```

```python
x.ndim
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=30)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred1=model.predict(x_train)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
mse_train=mean_squared_error(y_train,y_pred1)
print(mse_train)
```

```python
r2_train=r2_score(y_train,y_pred1)
print(r2_train)
```

```python
plt.scatter(x_train,y_train,color="r")
plt.plot(x_train,y_pred1,color='g')
plt.title("exp vs salary(train)")
plt.xlabel("Experience")
plt.ylabel("salary")
plt.show()
```

```python
mse_test=mean_squared_error(y_test,y_pred)
print(mse_test)
```

```python
r2_test=r2_score(y_test,y_pred)
print(r2_test)
```

```python
rmse_test=np.sqrt(mse_test)
mean_y=np.mean(y_test)

per=mean_y*0.1
print(rmse_test)
print(per)
```
